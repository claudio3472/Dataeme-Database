## Carregar csv

```python
import os
import pandas as pd
```

```python
df = pd.read_csv("produtos_realistas_sujos.csv")
df.head()
```

```python
print("Total rows: ", df.shape[0], " Total columns: ", df.shape[1])
print("Duplicated:", df["Ref Produto"].duplicated().sum())
print("Missing:", df["Preço Base"].isna().sum())
print("Unique:", df["Ref Produto"].nunique())
```

```python
display(
    df["Stock"]
    .value_counts()
    .sort_index()
    .reset_index()
    .rename(columns={
        "index": "Stock",
        "Stock": "Count"
    })
)

df["Stock"] = (df["Stock"].astype(str).str.replace(" unidades", "", regex=False))

display(
    df["Stock"]
    .value_counts()
    .sort_index()
    .reset_index()
    .rename(columns={
        "index": "Stock",
        "Stock": "Count"
    })
)
```

```python
print("Missing:", df["Ref Produto"].isna().sum())
print("Duplicated:", df["Ref Produto"].duplicated().sum())
print("Unique:", df["Ref Produto"].unique())
```

```python
nan_index = df[df["Ref Produto"].isna()].index[0]

df.loc[nan_index, "Ref Produto"] = "P016"
```

```python
print("Unique:", df["Ref Produto"].unique())
```

```python
print("Missing:", df["Family"].isna().sum())
print("Duplicated:", df["Family"].duplicated().sum())
print("Unique:", df["Family"].unique())
```

```python
df["Family"] = (df["Family"].astype(str).str.replace("ferramentas", "Ferramentas", regex=False))
```

```python
print("Missing:", df["Family"].isna().sum())
print("Duplicated:", df["Family"].duplicated().sum())
print("Unique:", df["Family"].unique())
```

```python
print("Missing:", df["SubFamily"].isna().sum())
print("Duplicated:", df["SubFamily"].duplicated().sum())
print("Unique:", df["SubFamily"].unique())
```

```python
for familia, df_fam in df.groupby("Family"):

    print(f"\n{'='*50}")
    print(f"FAMÍLIA: {familia}")
    print(f"{'='*50}")

    for subfamilia, df_sub in df_fam.groupby("SubFamily"):

        print(f"\n  Subfamília: {subfamilia}")

        for _, row in df_sub.iterrows():
            print(
                f"    {row['Ref Produto']} | "
                f"{row['Product Name']} | "
                f"{row['Barcode']} | "
                f"{row['Color']}"
            )
```

```python
df = df.drop_duplicates(subset=["Product Name"], keep="first")
```
